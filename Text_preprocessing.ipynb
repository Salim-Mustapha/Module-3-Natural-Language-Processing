# Code


```python
import pandas as pd
import re
import string

# Sample dataset
data = {
    "review": [
        "I LOVE this product!!! Visit https://example.com",
        "The service was AMAZING, and I am happy!",
        "This product is BAD... Do not buy it.",
        "The delivery was FAST!!! Check www.shop.com",
        "I really ENJOYED this purchase :)"
    ]
}

# Convert the data into a table
df = pd.DataFrame(data)

# Display the dataset
df
```

# Lower Casing

```python
# Make a copy of the original dataset
df["lowercase_review"] = df["review"].str.lower()

# Display the original and lowercase text
df[["review", "lowercase_review"]]
```

# Removal of URLs

```python
# Function to remove URLs from text
def remove_urls(text):
    text = re.sub(r"https?://\S+|www\.\S+", "", text)
    return text

# Apply URL removal to the lowercase text
df["no_urls"] = df["lowercase_review"].apply(remove_urls)

# Display the results
df[["lowercase_review", "no_urls"]]
```

# Removal of punctuation

```python
# Remove punctuation from the lowercase text
df["no_punctuation"] = df["lowercase_review"].apply(
    lambda text: text.translate(str.maketrans("", "", string.punctuation))
)

# Display the results
df[["lowercase_review", "no_punctuation"]]
```
